```python
import fasttext
```

## Split data to train/valid

head -n 693646 data/quora_wiki.txt > data/quora_wiki.train

head -n 173411 data/quora_wiki.txt > data/quora_wiki.valid

## Run fastetxt

```python
model = fasttext.train_supervised(input="../data/quora_wiki.train")
```

```python
model.predict("Which baking dish is best to bake a banana bread ?")
```

```python
model.test("../data/quora_wiki.valid", k=1)
```

```python
model = fasttext.train_supervised(input="../data/quora_wiki.train", epoch=25)
```

```python
model.test("../data/quora_wiki.valid", k=1)
```

```python
model = fasttext.train_supervised(input="../data/quora_wiki.train", lr=1.0, epoch=25)
```

```python
model.test("../data/quora_wiki.valid", k=1)
```

```python
model = fasttext.train_supervised(input="../data/quora_wiki.train", lr=1.0, epoch=25, wordNgrams=2)
```

```python
model.test("../data/quora_wiki.valid", k=1)
```
